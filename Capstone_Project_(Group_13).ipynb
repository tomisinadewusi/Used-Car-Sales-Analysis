```python
#Preview of car prices based on measures of central tendency
cars['price'].describe()
```

```python
#Distribution of car prices on a histogram
cars['price'].plot(kind='hist', bins=15)
plt.title('Distribution of Car Prices')
plt.xlabel('Price')
plt.show()
```

```python
# Mean price by fuel type

mean_prices = cars.groupby('fuelType')['price'].mean()

mean_prices
```

```python
#Plotting cars by fueltype against price on a bar plot

type_by_price=cars.groupby('fuelType')['price'].sum().sort_values(ascending=False)
type_by_price=pd.DataFrame(type_by_price)

type_by_price.plot(kind='bar',title='Fuel Type by Price')

for i, value in enumerate(type_by_price['price']):
  plt.text(i, value, str(value), ha='center', va='bottom')

plt.ylabel('Price')
plt.yticks([0, 10000000, 20000000, 30000000, 40000000], ["0M", "10M", "20M", "30M", "40M"])
plt.xticks(rotation=0)
plt.show()
```

```python
#Plotting cars by fueltype against tax on a bar plot

type_by_tax=cars.groupby('fuelType')['tax'].sum().sort_values(ascending=False)
type_by_tax=pd.DataFrame(type_by_tax)

type_by_tax.plot(kind='bar',title='Fuel Type by Tax')

for i, value in enumerate(type_by_tax['tax']):
  plt.text(i, value, str(value), ha='center', va='bottom')

plt.ylabel('Tax')
plt.yticks([0, 100000, 200000, 300000, 400000, 500000])
plt.xticks(rotation=0)

plt.show()
```

```python
#Plotting cars by fueltype against mpg on a bar plot

type_by_mpg=cars.groupby('fuelType')['mpg'].sum().sort_values(ascending=False)
type_by_mpg=pd.DataFrame(type_by_mpg)

type_by_mpg.plot(kind='bar',title='Fuel Type by MPG')

for i, value in enumerate(type_by_mpg['mpg']):
  plt.text(i, value, str(value), ha='center', va='bottom')

plt.ylabel('MPG')
plt.xticks(rotation=0)

plt.show()
```

```python
#Plotting cars by fueltype against mileage on a bar plot

type_by_mileage=cars.groupby('fuelType')['mileage'].sum().sort_values(ascending=False)
type_by_mileage=pd.DataFrame(type_by_mileage)

type_by_mileage.plot(kind='bar',title='Fuel Type by Mileage')

for i, value in enumerate(type_by_mileage['mileage']):
  plt.text(i, value, str(value), ha='center', va='bottom')

plt.ylabel('Mileage')
plt.yticks([0, 20000000, 40000000, 60000000, 80000000, 100000000], ["0M", "20M", "40M", "60M", "80M", "100M"])
plt.xticks(rotation=0)

plt.show()
```

```python
#To describe the dataset based on measures of central tendencies
cars.describe()
```

```python
#To generate the average price of hybrid cars compared to petrol cars (competition)
average_hybrid_price= cars[cars['fuelType'] == 'Hybrid']['price'].mean()

average_petrol_price= cars[cars['fuelType'] == 'Petrol']['price'].mean()

print(f"Average Price of a Hybrid car is: {round(average_hybrid_price)}")
print(f"Average Price of a Petrol car is: {round(average_petrol_price)}")
```

```python
#To count the number of hybrid cars sold

hybrid_count = cars[cars['fuelType'] == 'Hybrid'].shape[0]

print(f"Number of hybrid cars: {hybrid_count}")
```

```python
#To count the number of diesel cars sold

diesel_count = cars[cars['fuelType'] == 'Diesel'].shape[0]

print(f"Number of diesel cars: {diesel_count}")
```

```python
#To count the number of petrol cars sold

petrol_count = cars[cars['fuelType'] == 'Petrol'].shape[0]

print(f"Number of petrol cars: {petrol_count}")
```

```python
#To count the number of other cars sold

other_count = cars[cars['fuelType'] == 'Other'].shape[0]

print(f"Number of other cars: {other_count}")
```

```python
#To compute the total number of cars sold

total_cars = hybrid_count + diesel_count + petrol_count + other_count

print(f"Total number of cars: {total_cars}")
```

```python
#To compute a metric for calculating the percentage of hybrid cars sold

Hybrid_sales_share= (round((hybrid_count / total_cars) *100))
print(f"Percentage of Hybrid cars sold: {Hybrid_sales_share}%")
```

```python
#To compute the percentage of petrol cars sold

Petrol_sales_share= (round((petrol_count / total_cars) *100))
print(f"Percentage of Petrol cars sold: {Petrol_sales_share}%")
```

```python
#To compute the percentage of diesel cars sold

Diesel_sales_share= (round((diesel_count / total_cars) *100))
print(f"Percentage of Diesel cars sold: {Diesel_sales_share}%")
```

```python
#To compute the percentage of other cars sold

Other_sales_share= (round((other_count / total_cars) *100))
print(f"Percentage of Other cars sold: {Other_sales_share}%")
```

```python
# Assigning values to label and sizes for easy representation

labels = ['Hybrid', 'Petrol', 'Diesel', 'Other']
sizes = [Hybrid_sales_share, Petrol_sales_share, Diesel_sales_share, Other_sales_share]

# Create exploding pie chart
plt.figure()
plt.pie(sizes, labels=labels, autopct='%1.1f%%', explode = [0.03, 0.03, 0.08, 0.08,])

plt.title('Car Sales Distribution by Type')

plt.show()
```
